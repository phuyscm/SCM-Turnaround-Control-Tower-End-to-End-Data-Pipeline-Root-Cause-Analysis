```python
import pandas as pd
import numpy as np

# df = pd.read_csv('DataCoSupplyChainDataset.csv', encoding='latin1')

print("BẮT ĐẦU VALIDATION & CHỐT DATA MODEL...\n")

# =========================================================
# CHỐT 1: TẠO BẢNG DIM_GEOGRAPHY VÀ SURROGATE KEY
# =========================================================
# Tạo mã ID địa lý nhân tạo bằng cách ghép chuỗi (Ví dụ: Caguas_Puerto Rico)
df['Geo_Route_Id'] = df['Order City'] + "_" + df['Order Country']

geo_cols = ['Geo_Route_Id', 'Order City', 'Order State', 'Order Country', 'Order Region', 'Market']
dim_geography = df[geo_cols].drop_duplicates(subset=['Geo_Route_Id']).reset_index(drop=True)

print(f"Đã tạo Dim_Geography. Số tuyến đường độc nhất: {len(dim_geography)}")

# =========================================================
# CHỐT 3: ÉP KIỂU NGÀY THÁNG CHUẨN (Cực kỳ quan trọng cho BI)
# =========================================================
# Chuyển chuỗi thành đối tượng Datetime chuẩn của hệ thống
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])
df['shipping date (DateOrders)'] = pd.to_datetime(df['shipping date (DateOrders)'])

# =========================================================
# CHỐT 4: KIỂM TRA LOGIC KINH DOANH (Sanity Check)
# =========================================================
invalid_shipping_days = df[df['Days for shipping (real)'] < 0]
print(f"Số đơn hàng có ngày giao bị âm (Lỗi logic): {len(invalid_shipping_days)}")
# Nếu có, bạn có thể dùng code để xóa hoặc sửa đổi các dòng này.

# =========================================================
# CHỐT 2: KIỂM TRA TOÀN VẸN THAM CHIẾU (CHỐNG MỒ CÔI)
# =========================================================
# Cập nhật lại Fact_Orders có thêm Geo_Route_Id và định dạng ngày mới
fact_cols = ['Order Item Id', 'Order Id', 'Customer Id', 'Product Card Id', 'Geo_Route_Id',
             'order date (DateOrders)', 'Days for shipping (real)', 
             'Days for shipment (scheduled)', 'Late_delivery_risk', 'Sales']
fact_orders = df[fact_cols].copy()

# Tái tạo nhanh lại Dim_Customer và Dim_Product như bước trước
dim_customer = df[['Customer Id', 'Customer Fname', 'Customer Country']].drop_duplicates(subset=['Customer Id'])
dim_product = df[['Product Card Id', 'Product Name', 'Product Category Id']].drop_duplicates(subset=['Product Card Id'])

# Hàm kiểm tra mồ côi
def check_orphans(fact_df, dim_df, key_col):
    fact_keys = set(fact_df[key_col].unique())
    dim_keys = set(dim_df[key_col].unique())
    orphans = fact_keys - dim_keys
    return len(orphans)

print(f"Số Customer Id mồ côi: {check_orphans(fact_orders, dim_customer, 'Customer Id')}")
print(f"Số Product Id mồ côi: {check_orphans(fact_orders, dim_product, 'Product Card Id')}")
print(f"Số Geo Route Id mồ côi: {check_orphans(fact_orders, dim_geography, 'Geo_Route_Id')}")
# Mọi thứ in ra số 0 là bạn đã có một Data Model điểm 10 tuyệt đối!

# XUẤT FILE CUỐI CÙNG
dim_customer.to_csv('Dim_Customer.csv', index=False)
dim_product.to_csv('Dim_Product.csv', index=False)
dim_geography.to_csv('Dim_Geography.csv', index=False)
fact_orders.to_csv('Fact_Orders.csv', index=False)
print("\nĐÃ XUẤT XONG 4 FILE. SẴN SÀNG LÊN POWER BI!")
```

```python
import pandas as pd

print("BẮT ĐẦU CLEAN BUILD LẠI TOÀN BỘ PIPELINE...")

# 1. Tải lại Data gốc (Thay bằng đường dẫn của bạn nếu cần)
df = pd.read_csv('DataCoSupplyChainDataset.csv', encoding='latin1')

# 2. Tạo ID Địa lý (Để không bị lỗi đứt gãy Model ở bảng khác)
df['Geo_Route_Id'] = df['Order City'] + "_" + df['Order Country']

# 3. Tái tạo 3 Bảng Dim (Danh mục) chuẩn
dim_customer = df[['Customer Id', 'Customer Fname', 'Customer Lname', 'Customer Segment', 'Customer City', 'Customer Country']].drop_duplicates(subset=['Customer Id'])
dim_product = df[['Product Card Id', 'Product Name', 'Product Category Id', 'Category Name', 'Product Price']].drop_duplicates(subset=['Product Card Id'])
dim_geography = df[['Geo_Route_Id', 'Order City', 'Order State', 'Order Country', 'Order Region', 'Market']].drop_duplicates(subset=['Geo_Route_Id'])

# 4. Tái tạo Bảng Fact VỚI ĐỦ 13 CỘT (Đã thêm Shipping Mode, Quantity, Status)
fact_cols = [
    'Order Item Id', 'Order Id', 'Customer Id', 'Product Card Id', 'Geo_Route_Id',
    'order date (DateOrders)', 'Days for shipping (real)', 
    'Days for shipment (scheduled)', 'Late_delivery_risk', 'Sales',
    'Shipping Mode', 'Order Item Quantity', 'Order Status'  # <--- 3 CỘT CỨU TINH
]
fact_orders = df[fact_cols].copy()

# 5. Ép xuất file CSV đè lên file cũ
dim_customer.to_csv('Dim_Customer.csv', index=False)
dim_product.to_csv('Dim_Product.csv', index=False)
dim_geography.to_csv('Dim_Geography.csv', index=False)
fact_orders.to_csv('Fact_Orders.csv', index=False)

print("\nĐÃ XUẤT XONG 4 FILE MỚI!")
print("\nKIỂM TRA LẠI SỐ CỘT CỦA BẢNG FACT_ORDERS:")
# In ra thông tin để bạn check nóng luôn tại đây
print(fact_orders.info())
```
